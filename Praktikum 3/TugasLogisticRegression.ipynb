Import library yang diperlukan.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
```

Load dataset

```python
dataset_url = 'https://raw.githubusercontent.com/Lordyblade/IbrahimMousaDhani_2411532010_ML2526/refs/heads/main/Praktikum%202/Breast_cancer_data.csv'
data =  pd.read_csv(dataset_url)
data.head()
```

Memisahkan fitur (X) dan target (y)

```python
X = data.drop('diagnosis', axis=1)
y = data['diagnosis']
```

Split data menjadi data latih (70%) dan data uji (30%)

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.30, random_state=42)
```

SCALING DATASET

```python
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

MODEL TANPA SCALING

```python
logreg_unscaled = LogisticRegression(max_iter=1000, random_state=42)
logreg_unscaled.fit(X_train, y_train)
y_pred_prob_unscaled = logreg_unscaled.predict_proba(X_test)[:, 1]
```

PELATIHAN MODEL DENGAN SCALING

```python
logreg_scaled = LogisticRegression(random_state=42)
logreg_scaled.fit(X_train_scaled, y_train)
```

Mengambil nilai probabilitas prediksi kelas 1 (positif kanker)

```python
y_pred_prob_scaled = logreg_scaled.predict_proba(X_test_scaled)[:, 1]
```

PREDIKSI DAN EVALUASI BERBAGAI THRESHOLD

```python
def evaluate(y_true, y_prob, threshold, label):
    y_pred = (y_prob >= threshold).astype(int)
    print(f"- {label} (Threshold {threshold})")
    print(f"Accuracy: {accuracy_score(y_true, y_pred):.4f}")
    print("Classification Report:\n", classification_report(y_true, y_pred))
    print(" ")


# Cetak Evaluasi
evaluate(y_test, logreg_unscaled.predict_proba(X_test)[:, 1], 0.5, "MODEL TANPA SCALING")
evaluate(y_test, y_pred_prob_scaled, 0.5, "MODEL DENGAN SCALING") # Default
evaluate(y_test, y_pred_prob_scaled, 0.3, "MODEL DENGAN SCALING")
evaluate(y_test, y_pred_prob_scaled, 0.7, "MODEL DENGAN SCALING")
```
